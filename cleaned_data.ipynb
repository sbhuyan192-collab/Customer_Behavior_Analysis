```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\KIIT\OneDrive\Desktop\Customer_Shopping_Behavior_Analysis\data\raw\customer_shopping_behavior.csv"
)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform( lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.columns.tolist()

```

```python
df=df.rename(columns= {'purchase_amount_(usd)' : 'purchase_amount_usd'})
```

```python
df.columns
```

```python
#create a column age group
labels=['Young Adult','Adult','Middle_aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(20)
```

```python
print(df.columns.tolist())
```

```python
#purchase frequency days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

print(df[['frequency_of_purchases', 'purchase_frequency_days']].head(10))
```

```python
df.drop(columns=['age group'], inplace=True)
```
